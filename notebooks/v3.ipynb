```python
import requests

url = "https://api.openf1.org/v1/sessions"

params = {
    "year": 2026,
    "session_name": "Race",
}

response = requests.get(url, params=params)
response.raise_for_status()

data = response.json()

print(data)
```

```python
import pandas as pd


races=pd.DataFrame(data)
races['date_start']=pd.to_datetime(races['date_start'])

races.head()
```

```python
races.info()
```

```python
import pandas as pd

now = pd.Timestamp.now(tz='UTC')
filtered = races[races['date_start'] <= now]

filtered
```

```python
laps = requests.get(
    f"{BASE}/laps",
    params={
        "session_key": 11253,
        "driver_number": 1,
    },
).json()

laps_df = pd.DataFrame(laps)
```

```python
location = requests.get(
    f"{BASE}/location",
    params={
        "session_key": 11253,
        "driver_number": 1,
    },
).json()

location_df = pd.DataFrame(location)

location_df.head()
```

```python
location_df["date"] = pd.to_datetime(location_df["date"], utc=True)
```

```python
driver_number=1
lap_number=1


lap_start_time = laps_df[laps_df['driver_number']==driver_number].loc[lap_number]['date_start']
lap_start_time
```

```python
lap_pos=location_df[location_df["date"]<=lap_start_time]

lap_pos.head(5)
```

```python
plt.scatter(lap_pos['x'], lap_pos['y'], s=3)

plt.show()
```
