```python
import requests

url = "https://api.openf1.org/v1/sessions"

params = {
    "year": 2026,
    "session_name": "Race",
}

response = requests.get(url, params=params)
response.raise_for_status()

data = response.json()

print(data)
```

```python
import pandas as pd


races=pd.DataFrame(data)
races['date_start']=pd.to_datetime(races['date_start'])

races.head()
```

```python
races.info()
```

```python
import pandas as pd

now = pd.Timestamp.now(tz='UTC')
filtered = races[races['date_start'] <= now]

filtered
```
